MQTT Simulation Test

```python
from mqtt.simulated_mqtt import SimulatedMQTT

mqtt = SimulatedMQTT()

def receive_message(topic, message):
    print("Received:")
    print("Topic:", topic)
    print("Message:", message)

mqtt.subscribe("energy/load", receive_message)

mqtt.publish(
    "energy/load",
    {
        "load_kw": 4.5
    }
)
```
